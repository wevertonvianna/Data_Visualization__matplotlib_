```python
import pandas as pd
import matplotlib.pyplot as plt
import  seaborn as sns
import plotly.express as px
```

```python
raw = pd.read_csv('/content/imigrantes_canada.csv')
```

```python
raw.set_index('País',inplace= True)
```

```python
ameria_Sul =raw.query('Região == "América do Sul"')
```

```python
ameria_Sul
```

```python
cores = ['royalblue', 'orange', 'forestgreen', 'orchid', 'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy', 'teal', 'tomato']
```

```python
fig , ax = plt.subplots(figsize=(12,5))
ax.barh(ameria_Sul.index,ameria_Sul['Total'],color=cores)
ax.set_title('Imigração da America do sul para o Canadá\n1980 a 2013',loc='left',fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Numero de Imigrantes',fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
plt.show()
```

```python
america_sul_sort= ameria_Sul.sort_values('Total',ascending=True)
```

```python
america_sul_sort
```

```python
fig , ax = plt.subplots(figsize=(12,5))
ax.barh(america_sul_sort.index,america_sul_sort['Total'],color=cores)
ax.set_title('Imigração da America do sul para o Canadá\n1980 a 2013',loc='left',fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Numero de Imigrantes',fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
plt.show()
```

```python
cores=[]

for pais in america_sul_sort.index:
  if pais == 'Brasil':
    cores.append('royalblue')
  else:
    cores.append('silver')


fig , ax = plt.subplots(figsize=(12,5))
ax.barh(america_sul_sort.index,america_sul_sort['Total'],color=cores)
ax.set_title('America do sul: Brasil foi o quarto pais com mais  imigrantes \npara o Canadá mo período 1980 a 2013',loc='left',fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Numero de Imigrantes',fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
plt.show()
```

```python
cores=[]

for pais in america_sul_sort.index:
  if pais == 'Brasil':
    cores.append('royalblue')
  else:
    cores.append('silver')


fig , ax = plt.subplots(figsize=(12,5))
ax.barh(america_sul_sort.index,america_sul_sort['Total'],color=cores)
ax.set_title('America do sul: Brasil foi o quarto pais com mais  imigrantes \npara o Canadá mo período 1980 a 2013',loc='left',fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Numero de Imigrantes',fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

for i, v in enumerate(america_sul_sort['Total']):
  ax.text(v +20 ,i,str(v),color='black',fontsize=10,ha='left',va='center')

ax.set_frame_on(False)
ax.get_xaxis().set_visible(False)
ax.tick_params(axis='both',which='both',length=0)
#salvando
fig.savefig('imigração_america_sul.png',transparent=False,dpi=1200,bbox_inches='tight')
plt.show()
```

```python
def gerar_grafico_paleta(palette):
    fig ,ax = plt.subplots(figsize=(10,5))
    ax =sns.barplot(data=top_10, y=top_10.index,x='Total',orient='h',palette=palette)
    sns.set_theme(style="white", context="talk")
    ax.set_title('Top 10 países maior imigração para o Canadá\n1980 a 2013',loc='left',fontsize=18)
    ax.set_xlabel('Numero de Imigrantes',fontsize=14)
    ax.set_ylabel('')
```
